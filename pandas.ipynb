```python
data={
    'Employee_ID': ['E101','E102','E103','E104','E105','E106'],
    'Hours_Worked':[42,38,None,45,40,35],
    'Projects_Completed':[5,7,4,None,6,3],
    'Satisfaction_Score':[8.5,9.1,7.8,8.0,None,7.5]
}
```

```python
import pandas as pd
```

```python
df=pd.DataFrame(data)
```

```python
df
```

```python
df.describe()
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df
```

```python
df['Hours_Worked']=df['Hours_Worked'].fillna(df['Hours_Worked'].median())
```

```python
df.isnull().sum()
```

```python
df['Projects_Completed']=df['Projects_Completed'].fillna(df['Projects_Completed'].median())
```

```python
df['Satisfaction_Score']=df['Projects_Completed'].fillna(df['Projects_Completed'].median())
```

```python
df
```

```python
df['Efficiency_Ratio']=(df['Projects_Completed']/df['Hours_Worked'])*10
```

```python
df
```

```python
high_satisfaction=df[df["Satisfaction_Score"]>6.0]
```

```python
high_satisfaction
```

```python
import matplotlib.pyplot as plt
```

```python
plt.figure(figsize=(6,3))
bar_width=0
x=range(len(df))
plt.bar(x,df['Hours_Worked'],label='Hours_Worked')
plt.bar([p+bar_width for p in x],df['Projects_Completed'],label='Projects_Completed')
plt.xlabel("Employee Index")
plt.ylabel("Value")
plt.title("Comparision: Hours Worked and Projects Completed")
plt.legend()
plt.show()
```

```python
plt.figure(figsize=(6,4))
plt.scatter(df['Hours_Worked'],df['Satisfaction_Score'])
plt.xlabel("Hours_Worked")
plt.ylabel("Satisfaction Score")
plt.grid(True)
plt.show()
```

Q3. Matplotlib Visualization

```python
Semesters=["Sem1","Sem2","Sem3","Sem4"]
Average_CGPA_BA=[7.5,7.6,7.9,8.0]
Average_CGPA_BB=[7.2,7.6,7.9,8.0]
```

#PLot a grouped bar chart comapring batch A and Batch B for each semester
display cgpa values on the top of each bar


```python
plt.figure(figsize=(8,6))
x=range(0,len(Semesters))
bar_width=0.2
barA=plt.bar(Semesters,Average_CGPA_BA,label="Batch A",width=0.3)
barB=plt.bar([bar_width+p for p in x],Average_CGPA_BB,label="Batch B",width=0.3)
plt.xlabel("Semesters")
plt.ylabel("Average CGPA")
plt.title("Comparision between the Batch A & Batch B")
plt.legend()
# plt.grid(True)
plt.bar_label(barA)
plt.bar_label(barB)
plt.show()
```
