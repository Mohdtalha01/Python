Q2. frequency Distribution using tuples

```python
grades=('A','B','A','C','B','A','A','B','C')
# (("A",4),("B",3)..)
```

```python
print(grades.count('A'))
```

```python
L1=[]
L2=[]
for item in grades:
    # count=0
    if item not in L1:
        count=0
        # num=grades.count(item)
        for val in grades:
            if val==item:
                count=count+1
        L1.append(item)
        L2.append(count)
        
```

```python
print(L1)
print(L2)
```

```python
res=tuple(zip(L1,L2))
```

```python
res
```

Q3 Tuple data Aggregation

```python
sales=(
    ("Electronics",1000),
    ("Clothing",700),
    ("Electronics",1200),
    ("Toys",400),
    ("Clothing",600)
)
```

```python
for item in sales:
    print(item[0])
```

```python
sales[0][0]
```

```python
L1=[]
L2=[]

for item in sales:
    if item[0] not in L1:
        total=0      
        for val in sales:
            if item[0]==val[0]:
                total=total+val[1]
        L1.append(item[0])
        L2.append(total)

print(L1)
print(L2)
```

```python
res=tuple(zip(L1,L2))
print(res)
```
