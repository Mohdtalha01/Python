Q1. Statistical Computation from a File

Count, Mean, Median, Variance, Satndard deviation

```python
path="data.txt"
```

```python
f=open(path,encoding='utf')
```

```python
num_list=[]
for line in f:
    num_list.append(int(line)) 
    print(line)
```

```python
num_list
```

```python
def count_num(num_list):
    count=0
    for item in num_list:
        count=count+1
    return count
```

```python
print(f"The total numbers in the file is {count_num(num_list)}")
```

```python
def mean_num(num_list):
    sum=0
    for item in num_list:
        sum=sum+item
    mean_of_num=sum/count_num(num_list)
    return mean_of_num
    # print(f"The mean of the number given in the file is {mean_of_num}")
        
```

```python
print(f"The mean of the number given in the file is {mean_num(num_list)}")
```

```python
def selection_sort(arr):
    n = len(arr)
    for i in range(n):
        mini = i
        for j in range(i, n):
            if arr[j] < arr[mini]:
                mini = j
        arr[i], arr[mini] = arr[mini], arr[i]
```

```python
def median_num(num_list):
    temp=num_list.copy()
    selection_sort(temp)
    length=count_num(temp)
    if length%2!=0:
        ind=length//2
        med=temp[ind]
    else:
        ele1=temp[(length//2)-1]
        ind=((length//2) + 1)
        ele2=temp[ind-1]
        med=(ele1+ele2)/2
    return med
```

```python
median_num(num_list)
```

```python
def variance_num(num_list):
    sum=0
    # diff_list=[]
    length=count_num(num_list)
    temp_mean=mean_num(num_list)
    for item in num_list:
        diff=item - temp_mean
        # diff_list.append(diff)
        sum=sum+(diff**2)
        variance_of_num=sum/length-1
    return variance_of_num
```

```python
print(f"The variance of the numbers in the file is {variance_num(num_list)}")
```

```python
def std_num(num_list):
    temp=variance_num(num_list)
    variance_of_num=temp**(1/2)
    return variance_of_num
```

```python
print(f"The Standard deviation of the numbers in file is {std_num(num_list)}")
```
