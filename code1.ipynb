```python
class Employee:
    company='abc'
    list_of_emp=[]
    def __init__(self,name,age,salary,department):
        self.name=name
        Employee.list_of_emp.append(self)
        self.age=age
        self.salary=salary
        self.department=department
    
    def emp_info(self):
        print("The Information of the Employee : ")
        print(f"Name : {self.name}")
        print(f"Age :{self.age}")
        print(f"Salary :{self.salary}")
        return f"Department :{self.department}"
    
    def get_len(cls):
        return f"The number of Instances in list : {len(cls.list_of_emp)}"
    
```

```python
emp1=Employee("Jack",23,80000,"IT")
```

```python
print(emp1.emp_info())
```
