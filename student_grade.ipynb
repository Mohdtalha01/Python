Write a program in python to Evaluate Student's Grade

Input Student Name and marks in three subjects

```python
Student_name=input("Enter the name of the student- ")

sub_1=int(input("Enter the marks of First subject- "))
sub_2=int(input("Enter the marks of Second subject- "))
sub_3=int(input("Enter the marks of Third subject- "))

sum_marks=sub_1 + sub_2 + sub_3
Average=sum_marks//3
# print(Student_name)
if Average > 90:
    print(f"The Grade of {Student_name}  is A+")
elif Average > 80:
    print(f"The Grade of {Student_name}  is A")
elif Average > 70:
    print(f"The Grade of {Student_name}  is B")
elif Average > 60:
    print(f"The Grade of {Student_name}  is C")
elif Average > 50:
    print(f"The Grade of {Student_name}  is D")
else:
    print("Fail")
    
print("The Marks of the student are:-")
print(f"Subject_1 : {sub_1}")
print(f"Subject_2 : {sub_2}")
print(f"Subject_3 : {sub_3}")
```

```python
#Initialize the List_num that contains all the number given by user
List_num=[]
for i in range(10):
    #Taking input from the user
    num=int(input(f"Enter the {i+1} number -"))
    #Add the number into the list
    List_num.append(num)
print(List_num)
#Initialize the variable mini with values of first element in the list (assumption)
mini=List_num[0]
for i in range(10):
    #Iterate through list and find the number lesser than mini
    if(List_num[i]<mini):
        #If yes then assign that value to mini
        mini=List_num[i]
#Initialize the variable maximum with values of first element in the list (assumption)     
maximum=List_num[0]
for i in range(10):
    #Iterate through list and find the number greater than maximum
    if(List_num[i]>maximum):
        #If yes then assign that value to maximum
        maximum=List_num[i]

#finding the length of list using len
length = len(List_num)

print(f"The length of the list is {length}")
print(f"The maximum value in List is {maximum}")
print(f"The minimum value in List is {mini}")
for j in range(1,11):
    #print the list in reverse order using index
    print(List_num[10-j],sep=' ')
```

WAP in python to input a list of words and find the frequency of each unique word.
Display the result in a redeable format

```python
string_List=[]
freq={}
for i in range(5):
    inp=input(f"Enter the {i+1} word-")
    string_List.append(inp)
print(string_List)
for item in string_List:
    if item not in freq:
        freq[item]=1
    else:
        freq[item]+=1

print(freq)
```

WAP to remove all duplicate elements without using set() and maintain the original order

```python
def removedup(arr):
    # arr[2,2,3,4,5,6,7,8,2]
    print(f"The original array is -{arr}")
    n=len(arr)
    freq={}
    for item in arr:
        if item not in freq:
            freq[item]=1
        else:
            freq[item]+=1
    # print(freq)
    LL=[]
    for k,v in freq.items():
        LL.append(k)
    print(f"The List after removing the duplicates and preserve the order -{LL}")
```
